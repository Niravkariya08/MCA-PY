### 4.Write a program to find the sum of digits of a number using a while loop.

```python
#While
num1 = input("Enter Number :")
sum = 0
i = 0

while i < len(num1):
    typecast = int(num1[i])
    sum += typecast
    i += 1

print(sum)
```

```python
#For Loop
num1 = input("Enter Number :")
sum = 0
for i in num1:
    typecast = int(i)
    sum+=typecast

print(sum)
```

### 10.Write a program to generate a sequence of numbers using generator functions and yield keyword. 


```python
def count_up_to(max_value):
    count = 1
    while count <= max_value:
        yield count
        count += 1

counter = count_up_to(3)

print(next(counter)) 
print(next(counter)) 
print(next(counter))
```
