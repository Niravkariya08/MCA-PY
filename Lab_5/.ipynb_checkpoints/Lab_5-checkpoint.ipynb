```python
import pandas as pd
data = {
    'Name': ['Nirav', 'Pranav', 'Harsh', 'Ansh', 'Rahul'],
    'Age': [24, 27, 22, 32, 29],
    'Salary': [50000, 60000, 45000, 80000, 70000]
}
df = pd.DataFrame(data)

print("--- Original DataFrame ---")
print(df)

print("\n--- head(2): First 2 rows ---")
print(df.head(2))

print("\n--- tail(2): Last 2 rows ---")
print(df.tail(2))

print("\n--- describe(): Summary Statistics ---")
print(df.describe())

print("\n--- sort_values(): Sorted by Salary (Highest to Lowest) ---")
print(df.sort_values(by='Salary', ascending=False))
```

```python
import pandas as pd
import numpy as np

# 1. Create a DataFrame with missing data (NaN)
data = {
    'Name': ['Alice', 'Bob', 'Charlie', 'David', np.nan],
    'Age': [25, np.nan, 22, 32, 29],
    'Score': [85, 90, np.nan, 88, np.nan]
}
df = pd.DataFrame(data)

print("--- Original DataFrame (with missing values) ---")
print(df)

# 2. dropna(): Remove any rows that contain missing data
print("\n--- dropna(): Rows with missing data removed ---")
df_dropped = df.dropna()
print(df_dropped)

# 3. fillna(): Replace missing data with a specific value (e.g., 0)
print("\n--- fillna(0): Missing data filled with 0 ---")
df_filled_zero = df.fillna(0)
print(df_filled_zero)
```

```python
import matplotlib.pyplot as plt

# Create a figure to hold all 4 subplots
plt.figure(figsize=(10, 8))

# 1. Line Chart (plot)
plt.subplot(2, 2, 1) # (rows, columns, panel number)
plt.plot([1, 2, 3, 4, 5], [10, 25, 15, 30, 20], marker='o', color='blue')
plt.title("Line Chart")

# 2. Bar Chart (bar)
plt.subplot(2, 2, 2)
categories = ['A', 'B', 'C', 'D']
values = [5, 7, 3, 8]
plt.bar(categories, values, color='green')
plt.title("Bar Chart")

# 3. Histogram (hist)
plt.subplot(2, 2, 3)
# A list of numbers where '4' and '5' appear most frequently
data = [2, 3, 3, 4, 4, 4, 5, 5, 5, 6, 7, 8]
plt.hist(data, bins=5, color='orange', edgecolor='black')
plt.title("Histogram")

# 4. Pie Chart (pie)
plt.subplot(2, 2, 4)
labels = ['Apples', 'Bananas', 'Cherries']
sizes = [40, 35, 25]
plt.pie(sizes, labels=labels, autopct='%1.1f%%', colors=['red', 'yellow', 'pink'])
plt.title("Pie Chart")

# Adjust layout so labels don't overlap, and display the charts
plt.tight_layout()
plt.show()
```
