### 3. Write a program to perform arithmetic relational and logical operations using Python operators.

```python
num1 = int(input("Enter 1 Number : "))
num2 = int(input("Enter 2 Number : "))
num3 = int(input("Enter 3 Number : "))

print("\n--Arithmetic operation--")
print(f"Sum of {num1} + {num2} : {num1+num2}")
print(f"Sub of {num1} - {num2} : {num1-num2}")
print(f"Mul of {num1} * {num2} : {num1*num2}")
print(f"Div of {num1} / {num2} : {num1/num2}\n")

print("--Relational operation--")
print(f"{num1} is greater than {num2} : {num1>num2}")
print(f"{num2} is greater than {num1} : {num1<num2}")

print("\n--Logical operation--")
print(f"{num1} is greater than {num2} and {num1} is greater than {num3} : {num1 > num2 and num1 > num3}")
print(f"{num1} is greater than {num2} and {num1} is greater than {num3} : {num1 < num2 or num1 > num3}")
```

### 7. Write a program to create a dictionary and demonstrate dictionary methods and iteration. 


```python
user_information = {}

user_name  = input("Enter Your Name :")
user_age  = input("Enter Your Age :")
user_degree  = input("Enter Your Degree :")
user_information['name']=user_name
user_information['age']=user_age
user_information['degree']=user_degree
print("Full Dictionary = ",user_information)

user_information.pop('age')
print("After Pop Operation = ",user_information)

print("Dictionary Keys = ",user_information.keys())
print("Dictionary Values = " ,user_information.values())

user_information.clear()
print("After Clear Operation = ",user_information)
```

### 10.Write a program to demonstrate recursion using factorial or Fibonacci series.

```python
def factorial_recursive(n):
    if n < 0:
        raise ValueError("Factorial is not defined for negative numbers.")
    if n < 2:
        return 1
    
    return n * factorial_recursive(n - 1)

user_input = int(input('Enter Number :'))
print(factorial_recursive(user_input))
```
